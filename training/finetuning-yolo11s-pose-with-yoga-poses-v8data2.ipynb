```python
# 🚫 Disable all integrations that can break on Kaggle
import os
os.environ["WANDB_MODE"] = "disabled"      # disable Weights & Biases
os.environ["ULTRALYTICS_TB"] = "False"     # disable TensorBoard
os.environ["RAY_MODE"] = "disabled"        # disable Ray Tune callbacks
os.environ["RAY_DISABLE_IMPORT"] = "1"     # ensure Ray never loads internally
```

```python
from ultralytics import YOLO

model = YOLO('yolo11s-pose.pt')
model.train(
    data='/kaggle/input/yoga-poses-v8/Workout_Pose.v8i.yolov8/data.yaml',
    epochs=100,
    imgsz=640,
    batch=16,
    project='runs',
    name='yolo11s_pose_finetune',
    device=0
)
```

```python
from ultralytics import YOLO

model = YOLO("runs/yolo11s_pose_finetune/weights/best.pt")

results = model.predict(
    source="/kaggle/input/yoga-poses-v8/Workout_Pose.v8i.yolov8/test/images",
    project="runs",
    name="pose_test_predictions",
    save=True
)
```

```python
from IPython.display import Image, display
import os

folder = "runs/pose_test_predictions"
images = sorted([f for f in os.listdir(folder) if f.lower().endswith(('.jpg', '.png'))])

# Display first 5 predictions
for i, img_file in enumerate(images[:10]):
    display(Image(filename=os.path.join(folder, img_file), width=500))
```
